```python
import os
import torch

'''
ATENÇÃO:
Acesse o menu/submenu Editar => Configurações de Notebook.
Escolha a opção T4 GPU e clique em salvar.
'''


#####################
# MONTAGEM DO DRIVE #
#####################
from google.colab import drive
if not os.path.ismount('/content/drive'):
    print("Montando o Google Drive...")
    drive.mount('/content/drive')
else:
    print("O Google Drive já está montado.")

########################################################################
# CAMINHOS E VARIAVES GLOBAIS
########################################################################

caminho_yolov9_git = 'https://github.com/WongKinYiu/yolov9.git'
#pt = pre-trained ou pre-treinado
caminho_yolov9_c_pt_git = 'https://github.com/WongKinYiu/yolov9/releases/download/v0.1/yolov9-c.pt'
caminho_yolov9_e_pt_git = 'https://github.com/WongKinYiu/yolov9/releases/download/v0.1/yolov9-e.pt'

caminho_yolov9_local = '/content/drive/MyDrive/Colab Notebooks/VC/yolov9/'
caminho_yolov9_script_detect = caminho_yolov9_local + 'detect.py'
caminho_yolov9_script_train = caminho_yolov9_local + 'train_dual.py'
caminho_yolov9_c_pt_local = caminho_yolov9_local + 'yolov9-c.pt'
caminho_yolov9_e_pt_local = caminho_yolov9_local + 'yolov9-e.pt'
caminho_yolov9_yaml_custom = caminho_yolov9_local + 'models/detect/yolov9_custom.yaml'
caminho_yolov9_hyp = caminho_yolov9_local + 'data/hyps/hyp.scratch-high.yaml'
caminho_yolov9_e_custom_local = caminho_yolov9_local + 'runs/train/exp/weights/best.pt'
dispositivo = None

###########################
# CARREGAMENTO DE MÓDULOS #
###########################

# Caminho dos módulos do projeto armazenados no Google Drive
caminho_modulos = '/content/drive/MyDrive/Colab Notebooks/VC/Deteccao_Yolov9_Custom/modulos'
from pydoc import importfile
# Importação de módulos Python
modulo_utils = importfile(caminho_modulos+'/utils.py')

#####################################
# INICIALIZACAO
#####################################

# Clonar repositório Yolov9 do GitHub para o Google Drive
modulo_utils.clonar_yolov9(caminho_yolov9_git,caminho_yolov9_local)

# Obter informações sobre os dispositivos
# GPU Nvidia instalados no sistema
!nvidia-smi

# Verifica disponibilidade de GPU
dispositivo = modulo_utils.verificar_gpu()

# Baixar modelos treinados

# Modelo compacto
modulo_utils.baixar_modelo_yolov9_treinado(
    caminho_yolov9_c_pt_git,
    caminho_yolov9_local
)

# Modelo estendido
modulo_utils.baixar_modelo_yolov9_treinado(
    caminho_yolov9_e_pt_git,
    caminho_yolov9_local
)


########################################################################
# MENU PRINCIPAL
 ########################################################################

def menu():
    while True:
        print("\n")
        print("Selecione uma opção:")
        print("1. Treinar com um dataset rotulado")
        print("2. Testar o modelo customizado")
        print("3. Sair")
        print("\n")

        escolha = input("Digite o número da sua escolha: ")

        if escolha == '1':
            # Estabelece parâmetros
            script_file = caminho_yolov9_script_train
            batch = 4
            epochs = 10
            dataset = caminho_yolov9_local + '/data/images/wood_log/data.yaml'
            data_yaml_file = caminho_yolov9_yaml_custom
            # Prepara o comando shell que será executado para o treinamento
            comando_shell = (
                f'python "{script_file}" '
                f'--workers 8 --device {dispositivo} --batch {batch} --epochs {epochs} --img 640 '
                f'--data "{dataset}" '
                f'--cfg "{caminho_yolov9_yaml_custom}" '
                f'--weights "{caminho_yolov9_e_pt_local}" '
                f'--hyp "{caminho_yolov9_hyp}"'
            )
            # Executa o comando shell
            !{comando_shell}

        elif escolha == '2':
            caminho_imagem = os.path.join(caminho_yolov9_local, 'data/images/wood_log/test/images/294552602_381095794167019_4592774562126642538_n_jpg.rf.748602f772f2f7158382c85b2c7bfb36.jpg')
            !python "{caminho_yolov9_script_detect}" --weights "{caminho_yolov9_e_custom_local}" --source "{caminho_imagem}" --device {dispositivo}
        elif escolha == '3':
            print("Execução finalizada!")
            break
        else:
            print("Escolha inválida. Por favor, selecione uma opção válida.")

# Chamada do menu para que o usuário faça a escolha
menu()


################ FIM DO SCRIPT ########################
```
