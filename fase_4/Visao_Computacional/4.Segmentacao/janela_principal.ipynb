```python
# @title
import ipywidgets as widgets
from IPython.display import display, HTML
import matplotlib.pyplot as plt
import numpy as np
import sys
import os
import colorsys

#####################
# MONTAGEM DO DRIVE #
#####################
from google.colab import drive
if not os.path.ismount('/content/drive'):
    print("Montando o Google Drive...")
    drive.mount('/content/drive')

###########################
# CARREGAMENTO DE MÓDULOS #
###########################

# Caminho dos módulos do projeto armazenados no Google Drive
caminho_modulos = '/content/drive/MyDrive/Colab Notebooks/VC/Segmentacao/modulos'
from pydoc import importfile
# Importa um módulo Python utilizando o caminho de um arquivo Python
filtros = importfile(caminho_modulos+'/filtros.py')
aprimoramentos = importfile(caminho_modulos+'/aprimoramentos.py')
segmentacao = importfile(caminho_modulos+'/segmentacao.py')

########################################################################
# CAMINHO DAS IMAGENS E VARIAVEL GLOBAL PARA A IMAGEM CORRENTE (ATUAL) #
########################################################################

# Caminho das imagens do projeto armazenadas no Google Drive
caminho_imagens = '/content/drive/MyDrive/Colab Notebooks/VC/Segmentacao/imagens'

# Variáveis globais para armazenar as imagens
imagem_original = None
imagem_corrente = None
imagem_segmentada = None
imagem_mascara = None


##################################################################
# FUNÇÕES CALLBACK ASSOCIADAS AOS ITENS DA INTERFACE DO  USUÁRIO #
##################################################################

# Função para carregar e exibir a imagem selecionada
def abrir_imagem(b):
    fechar_imagem(b)
    fechar_histograma(b)
    fechar_segmentacao(b)
    caminho_arquivo = os.path.join(caminho_imagens, dropdown_arquivos.value)
    global imagem_original
    global imagem_corrente
    imagem_original = filtros.read_image(caminho_arquivo)
    imagem_corrente = imagem_original
    if imagem_corrente is not None:
        exibir_imagem(imagem_corrente)
    else:
        print('Erro ao carregar a imagem.')

# Função para fechar a imagem corrente
def fechar_imagem(b):
    image_output.clear_output()
    global imagem_corrente, imagem_original
    imagem_corrente = None
    imagem_original = None

# Função para redimensionar a imagem corrente
def redimensionar(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.resize_image(imagem_corrente,
                                             redimensionar_largura.value,
                                             redimensionar_altura.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para converter para escala de cinza a imagem corrente
def escala_cinza(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.grayscale_image(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o filtro media na imagem corrente
def filtro_media(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.average_filter(imagem_corrente, slider_media.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o filtro gaussiano na imagem corrente
def filtro_gaussiano(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.gaussian_filter(imagem_corrente, slider_gaussiano.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o filtro mediana na imagem corrente
def filtro_mediana(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.median_filter(imagem_corrente, slider_mediana.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o filtro sobel na imagem corrente
def filtro_sobel(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.sobel_filter(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o filtro laplaciano na imagem corrente
def filtro_laplaciano(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.laplacian_filter(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')


# Função para aplicar o filtro highboost na imagem corrente
def filtro_highboost(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.highboost_filter(imagem_corrente, slider_highboost.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o ruído sal e pimenta na imagem corrente
def ruido_salpimenta(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = filtros.salt_and_pepper_noise(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')


# Criar um widget Output para exibir o histograma com tamanho limitado
hist_output = widgets.Output(layout=widgets.Layout(max_width='500px',
                                                   max_height='500px'))

# Função para exibir o histograma na tela
def exibir_histograma(b):
    with hist_output:
        hist_output.clear_output(wait=True)
        aprimoramentos.show_histogram(imagem_corrente)


# Função para converter para negativo a imagem corrente
def negativo(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = aprimoramentos.negative_image(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')


# Função para aplicar a transf logaritmica na imagem corrente
def transf_logaritmica(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = aprimoramentos.log_transform(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar a correcao gamma na imagem corrente
def correcao_gamma(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = aprimoramentos.gamma_correction(imagem_corrente, slider_gamma.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar o ajuste de  contraste na imagem corrente
def ajuste_contraste(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = aprimoramentos.contrast_stretch(imagem_corrente,slider_ajuste_maior.value,slider_ajuste_menor.value)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar a equalizacao de histograma na imagem corrente
def equalizacao_histograma(b):
    global imagem_corrente
    if imagem_corrente is not None:
        imagem_corrente = aprimoramentos.histogram_equalization(imagem_corrente)
        exibir_imagem(imagem_corrente)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Output widget para exibir uma imagem segmentada na interface
segmentation_output = widgets.Output(layout={'border': '1px solid black'})

# Função para exibir o resultado da segmentação
def exibir_segmentacao(b):
    with segmentation_output:
        segmentation_output.clear_output(wait=True) # Limpar a saída anterior, mas espera até o novo conteúdo estar pronto
        if imagem_segmentada is not None:
          if button_mascara.description == "Mostrar Máscara":
              button_mascara.description = "Mostrar Segmentação"
              imagem = imagem_mascara
          else:
              imagem = imagem_segmentada
              button_mascara.description = "Mostrar Máscara"
          if imagem is not None:
              dpi = 100  # Define a densidade de pixels por polegada
              height, width = imagem.shape[:2]  # Obtém a altura e a largura da imagem
              # a imagem deve ser exibida em seu tamanho real sem redimensionamento
              fig, ax = plt.subplots(figsize=(width / dpi, height / dpi), dpi=dpi)
              ax.imshow(imagem, cmap='gray')
              ax.axis('off')
              plt.show()

# Função para aplicar a segmentação por cor
def segmentacao_cor(b):
    global imagem_corrente, imagem_segmentada, imagem_mascara
    if imagem_corrente is not None:
        imagem_segmentada, imagem_mascara = segmentacao.color_segmentation(imagem_corrente,
                                                                           h_min.value, h_max.value,
                                                                           s_min.value, s_max.value,
                                                                           v_min.value, v_max.value)
        exibir_segmentacao(b)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar a segmentação por limiarização
def segmentacao_limiar(b):
    global imagem_corrente, imagem_segmentada, imagem_mascara
    if imagem_corrente is not None:
        imagem_segmentada, imagem_mascara = segmentacao.threshold(imagem_corrente,slider_limiar.value)
        exibir_segmentacao(b)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar a segmentação por limiarização
def segmentacao_otsu(b):
    global imagem_corrente, imagem_segmentada, imagem_mascara
    if imagem_corrente is not None:
        imagem_segmentada, imagem_mascara, t = segmentacao.otsu_threshold(imagem_corrente)
        texto_otsu.value = f'Limiar de Otsu = {t}'
        box_texto_otsu.layout.display = 'flex'  # exibir texto
        exibir_segmentacao(b)
    else:
        with image_output:
            print('Não existe imagem carregada.')


# Função para aplicar a segmentação por limiarização
def segmentacao_canny(b):
    global imagem_corrente, imagem_segmentada, imagem_mascara
    if imagem_corrente is not None:
        imagem_segmentada, imagem_mascara, n = segmentacao.canny(imagem_corrente,slider_canny.value)
        texto_canny.value = f'Número de regiões = {n}'
        box_texto_canny.layout.display = 'flex'  # exibir texto
        exibir_segmentacao(b)
    else:
        with image_output:
            print('Não existe imagem carregada.')

# Função para aplicar a segmentação por limiarização
def segmentacao_watershed(b):
    global imagem_corrente, imagem_segmentada, imagem_mascara
    if imagem_corrente is not None:
        imagem_segmentada, imagem_mascara = segmentacao.watershed_segmentation(imagem_corrente)
        exibir_segmentacao(b)
    else:
        with image_output:
            print('Não existe imagem carregada.')


###########################################
# FUNÇÕES AUXILIARES ÀS FUNÇÕES CALLBACKS #
###########################################

# Função para listar arquivos de imagem no diretório atual
def listar_arquivos(diretorio):
    arquivos = os.listdir(diretorio)
    arquivos_img = [arquivo for arquivo in arquivos if arquivo.lower().endswith(('png', 'jpg', 'jpeg'))]
    return arquivos_img

# Função para exibir uma imagem utilizando a matplotlib
def exibir_imagem(image):
    with image_output: # Widget para a interface gráfica web
        image_output.clear_output(wait=True)  # Limpar a saída anterior, mas espera até o novo conteúdo estar pronto
        dpi = 100  # Define a densidade de pixels por polegada
        height, width = image.shape[:2]  # Obtém a altura e a largura da imagem
        # a imagem deve ser exibida em seu tamanho real sem redimensionamento
        fig, ax = plt.subplots(figsize=(width / dpi, height / dpi), dpi=dpi)
        ax.imshow(image, cmap='gray')
        ax.axis('off')
        plt.show()

def restaurar_imagem(b):
    if imagem_original is not None:
        global imagem_corrente
        imagem_corrente = imagem_original
        exibir_imagem(imagem_corrente)  # Restaura a imagem original

def fechar_histograma(b):
    hist_output.clear_output()

def fechar_segmentacao(b):
    global imagem_segmentada, imagem_mascara
    segmentation_output.clear_output()
    imagem_segmentada = None
    imagem_mascara = None
    button_mascara.description = 'Mostrar Segmentação'
    box_texto_otsu.layout.display = 'none'  # ocultar texto
    box_texto_canny.layout.display = 'none'  # exibir texto

#################################
# ITENS DA INTERFACE DO USUÁRIO #
#################################

#############
# CABEÇALHO #
#############

# Header
header = widgets.HTML('<font color="white" face="sans-serif"><center><h1>Pré-processamento e Segmentação de Imagens</h1></center></font>',
                      layout=widgets.Layout(height='auto'))

# Botão para restaurar a imagem (será utilizado em várias abas)
button_restaurar = widgets.Button(description='Restaurar')
# Função atribuída ao botão restaurar
button_restaurar.on_click(restaurar_imagem)
# Botão para exibir máscara (será utilizado em várias abas)
button_mascara = widgets.Button(description='Mostrar Segmentação')
# Atribuir função ao botão
button_mascara.on_click(exibir_segmentacao)

###################
# ABAS PRINCIPAIS #
###################

# Criar abas
tab = widgets.Tab(
    layout=widgets.Layout(min_width='700px', min_height='50px') # Definir tamanho mínimo
)

c_base = int("1F534",base=16)

# Definir os títulos de cada aba
tab.set_title(0, 'ARQUIVO')
tab.set_title(1, 'IMAGEM')
tab.set_title(2, 'FILTROS')
tab.set_title(3, 'APRIMORAMENTOS')
tab.set_title(4, 'SEGMENTAÇÃO')

# Output widget para exibir uma imagem na interface
image_output = widgets.Output(layout={'border': '1px solid black'})

# HBox para organizar image_output, hist_output e segmentation lado a lado
outputs_box = widgets.HBox([image_output, hist_output, segmentation_output], layout=widgets.Layout(align_items='center'))

# GridBox para organizar o cabeçalho, as abas e a imagem de saída
grid = widgets.GridBox(children=[header, tab, outputs_box],
               layout=widgets.Layout(
                   grid_template_rows='auto auto auto',
                   grid_template_columns='1fr',  #uma  coluna flexível para ocupar o espaço disponível.
                   justify_items='center',
                   align_items='center',
                   grid_gap='10px'
               ))

# Contêiner pai para centralizar o GridBox
container = widgets.Box([grid], layout=widgets.Layout(display='flex', justify_content='center', align_items='center'))


# Exibir o contêiner pai
display(container)


########################
# CONTEUDO ABA ARQUIVO #
########################

# Dropdown para selecionar um arquivo de imagem
dropdown_arquivos = widgets.Dropdown(
    options=listar_arquivos(caminho_imagens),
    description='Imagem:',
)
# Botões
button_abrir = widgets.Button(description='Abrir')
button_fechar = widgets.Button(description='Fechar')
# Atribuir funções aos botões
button_abrir.on_click(abrir_imagem)
button_fechar.on_click(fechar_imagem)

########################
# CONTEUDO ABA IMAGEM  #
########################

########################
# ITENS REDIMENSIONAR  #
########################

# Botão
button_redimensionar = widgets.Button(description='Aplicar')
# Atribuir função aos botão
button_redimensionar.on_click(redimensionar)
# Caixa de texto para redimensionar imagem
redimensionar_largura = widgets.BoundedIntText(value=50,
                                      min=0,
                                      max=200,
                                      step=1,
                                      description='Largura (%):',
                                      disabled=False,
                                      layout=widgets.Layout(width='150px')
                                  )
redimensionar_altura = widgets.BoundedIntText(value=50,
                                      min=0,
                                      max=200,
                                      step=1,
                                      description='Altura (%):',
                                      disabled=False,
                                     layout=widgets.Layout(width='150px')
                                  )

# Box para os itens de Redimensionar
box_redimensionar = widgets.Box([redimensionar_largura, redimensionar_altura, button_redimensionar, button_restaurar],
                                layout=widgets.Layout(justify_content='center',gap='1px',margin='0px'))

##########################
# ITENS ESCALA DE CINZA  #
##########################

# Botão
button_escala_cinza = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_escala_cinza.on_click(escala_cinza)
# Box para os itens de Escala de Cinza
box_escala_cinza = widgets.Box([button_escala_cinza, button_restaurar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

# Criar subabas para a aba Imagem
subtab_imagem = widgets.Tab(
    children=[
        widgets.VBox([box_redimensionar]),
        widgets.VBox([box_escala_cinza])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_imagem.set_title(0, 'Redimensionar')
subtab_imagem.set_title(1, 'Escala de Cinza')


#########################
# CONTEUDO ABA FILTROS  #
#########################

#######################
#    ITENS MEDIA       #
#######################

# Slider
slider_media = widgets.IntSlider(value=3, min=3, max=11, step=2)
# Botão
button_media = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_media.on_click(filtro_media)


#######################
#    ITENS GAUSSIANO  #
#######################

# Slider
slider_gaussiano = widgets.IntSlider(value=3, min=3, max=11, step=2)
# Botão
button_gaussiano = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_gaussiano.on_click(filtro_gaussiano)

#######################
#    ITENS MEDIANA  #
#######################

# Slider
slider_mediana = widgets.IntSlider(value=3, min=3, max=11, step=2)
# Botão
button_mediana = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_mediana.on_click(filtro_mediana)

#####################
# ITENS SUAVIZACAO  #
#####################

# Boxes para os itens de Suavização
box_media = widgets.Box([slider_media, button_media, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))
box_gaussiano = widgets.Box([slider_gaussiano, button_gaussiano, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))
box_mediana = widgets.Box([slider_mediana, button_mediana, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))

# Criar subabas com os boxes de Suavização
subtab_suavizacao = widgets.Tab(
    children=[
        widgets.VBox([box_media]),
        widgets.VBox([box_gaussiano]),
        widgets.VBox([box_mediana])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_suavizacao.set_title(0, 'Média')
subtab_suavizacao.set_title(1, 'Gaussiano')
subtab_suavizacao.set_title(2, 'Mediana')

#####################
# ITENS AGUÇAMENTO  #
#####################


#######################
#    ITENS SOBEL      #
#######################

# Botão
button_sobel = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_sobel.on_click(filtro_sobel)


########################
#    ITENS LAPLACIANO  #
########################

# Botão
button_laplaciano = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_laplaciano.on_click(filtro_laplaciano)

########################
#    ITENS HIGH BOOST  #
########################

# Slider
slider_highboost = widgets.FloatSlider(value=0.1, min=0.1, max=1.0, step=0.1)
# Botão
button_highboost = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_highboost.on_click(filtro_highboost)

# Boxes para os itens de Aguçamento
box_sobel = widgets.Box([button_sobel, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))
box_laplaciano = widgets.Box([button_laplaciano, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))
box_highboost = widgets.Box([slider_highboost,button_highboost, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))

# Criar subabas com os boxes de Aguçamento
subtab_agucamento = widgets.Tab(
    children=[
        widgets.VBox([box_sobel]),
        widgets.VBox([box_laplaciano]),
        widgets.VBox([box_highboost])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_agucamento.set_title(0, 'Sobel')
subtab_agucamento.set_title(1, 'Laplaciano')
subtab_agucamento.set_title(2, 'High Boost')

################
# ITENS RUÍDO  #
################

###########################
#    ITENS SAL E PIMENTA  #
###########################

# Botão
button_salpimenta = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_salpimenta.on_click(ruido_salpimenta)

# Box para os itens de Ruído
box_salpimenta = widgets.Box([button_salpimenta, button_restaurar], layout=widgets.Layout(justify_content='center',gap='1px'))

# Criar subabas com os box de Ruído
subtab_ruido = widgets.Tab(
    children=[
        widgets.VBox([box_salpimenta]),
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_ruido.set_title(0, 'Sal e pimenta')

# Criar subabas para a aba Filtros
subtab_filtros = widgets.Tab(
    children=[
        widgets.VBox([subtab_suavizacao]),
        widgets.VBox([subtab_agucamento]),
        widgets.VBox([subtab_ruido])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_filtros.set_title(0, 'Suavização')
subtab_filtros.set_title(1, 'Aguçamento')
subtab_filtros.set_title(2, 'Ruído')

##########################
# ABA  APRIMORAMENTOS    #
##########################

##########################
# ITENS NEGATIVO         #
##########################

# Botão
button_negativo = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_negativo.on_click(negativo)
# Box para os itens de Negativo
box_negativo = widgets.Box([button_negativo, button_restaurar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

##########################
# ITENS BRILHO/CONT      #
##########################

##########################
# ITENS TRANSF. LOG.     #
##########################

# Botão
button_transf_log = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_transf_log.on_click(transf_logaritmica)
# Box para os itens de Transf Log
box_transf_log = widgets.Box([button_transf_log , button_restaurar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

##########################
# ITENS CORREÇÃO GAMMA     #
##########################

# Slider
slider_gamma = widgets.FloatSlider(value=1.5, min=0.1, max=3.0, step=0.1)
# Botão
button_gamma = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_gamma.on_click(correcao_gamma)
# Box para os itens de Transf Log
box_gamma = widgets.Box([slider_gamma, button_gamma, button_restaurar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))
##########################
# ITENS AJUSTE DE CONT   #
##########################

# Slider
slider_ajuste_maior = widgets.IntSlider(value=255, min=0, max=255, step=1, description='Nível max:')
# Slider
slider_ajuste_menor = widgets.IntSlider(value=0, min=0, max=255, step=1, description='Nível min:')
# Botão
button_ajuste_cont = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_ajuste_cont.on_click(ajuste_contraste)
# Box para os itens de Transf Log
box_ajuste_cont = widgets.Box([slider_ajuste_maior, slider_ajuste_menor, button_ajuste_cont, button_restaurar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

##########################
# ITENS EQ. HISTOGRAMA   #
##########################


# Botão
button_equal_hist = widgets.Button(description='Aplicar')
# Atribuir função ao botão
button_equal_hist.on_click(equalizacao_histograma)
# Box para os itens de Transf Log
box_equal_hist = widgets.Box([button_equal_hist, button_restaurar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

# Criar subabas com os box de Brilho e Cont
subtab_brilho = widgets.Tab(
    children=[
        widgets.VBox([box_transf_log]),
        widgets.VBox([box_gamma]),
        widgets.VBox([box_ajuste_cont]),
        widgets.VBox([box_equal_hist])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_brilho.set_title(0, 'Transform.  Logarítmica')
subtab_brilho.set_title(1, 'Correção Gamma')
subtab_brilho.set_title(2, 'Ajuste de Contraste')
subtab_brilho.set_title(3, 'Equalização de Histograma')


##########################
# ITENS HISTOGRAMA       #
##########################

# Botão
button_histograma = widgets.Button(description='Exibir')
# Atribuir função ao botão
button_histograma.on_click(exibir_histograma)
# Botão
button_fechar_histograma = widgets.Button(description='Fechar')
# Atribuir função ao botão
button_fechar_histograma.on_click(fechar_histograma)
# Box para os itens de Histograma
box_histograma = widgets.Box([button_histograma, button_fechar_histograma],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

# Criar subabas para a aba Aprimoramentos
subtab_aprimoramentos = widgets.Tab(
    children=[widgets.VBox([box_negativo]),
              widgets.VBox([box_histograma]),
              widgets.VBox([subtab_brilho])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_aprimoramentos.set_title(0, 'Negativo')
subtab_aprimoramentos.set_title(1, 'Histograma')
subtab_aprimoramentos.set_title(2, 'Brilho e Contraste')

##########################
# ABA  SEGMENTAÇÃO       #
##########################

##########################
# ITENS SEGMENTACAO COR  #
##########################

# Define os widgets para seleção do intervalo de cores HSV
h_min = widgets.IntSlider(min=0, max=179, step=1, value=166, description='H min')
h_max = widgets.IntSlider(min=0, max=179, step=1, value=179, description='H max')
s_min = widgets.IntSlider(min=0, max=255, step=1, value=50, description='S min')
s_max = widgets.IntSlider(min=0, max=255, step=1, value=255, description='S max')
v_min = widgets.IntSlider(min=0, max=255, step=1, value=50, description='V min')
v_max = widgets.IntSlider(min=0, max=255, step=1, value=255, description='V max')
# Seletor de cores
color_picker = widgets.ColorPicker(description='Escolha uma cor:',
                                   layout=widgets.Layout(width='200px'),
                                   style={'description_width': 'initial'})

def on_color_change(change):
        # Converte a cor selecionada para HSV e atualiza os sliders para H
        r, g, b = [int(change['new'][i:i+2], 16) for i in (1, 3, 5)]
        hsv = colorsys.rgb_to_hsv(r/255.0, g/255.0, b/255.0)
        h, s, v = int(hsv[0]*179), int(hsv[1]*255), int(hsv[2]*255)
        h_min.value = max(0, h-10)
        h_max.value = min(179, h+10)

color_picker.observe(on_color_change, names='value')
# Botão para segmentar
button_segmentar_cor = widgets.Button(description='Segmentar')
# Atribuir função ao botão
button_segmentar_cor.on_click(segmentacao_cor)
# Botão fechar segmentação
button_fechar_segmentacao = widgets.Button(description='Fechar')
# Atribuir função ao botão
button_fechar_segmentacao.on_click(fechar_segmentacao)

# Box para organizar os controles hsv min
box_hsv_min = widgets.Box([h_min,s_min,v_min],
                          layout=widgets.Layout(justify_content='center',margin='0px'))
# Box para organizar os controles hsv max
box_hsv_max = widgets.Box([h_max,s_max,v_max],
                          layout=widgets.Layout(justify_content='center',margin='0px'))
# Box para organizar os controles de cor e segmentar
box_cor_seg = widgets.Box([color_picker,button_segmentar_cor,button_mascara,button_fechar_segmentacao],
                          layout=widgets.Layout(justify_content='center',margin='0px'))
# Box para organizar todos os controles da aba segmentação
box_hsv = widgets.VBox([box_hsv_min, box_hsv_max,box_cor_seg],
                       layout=widgets.Layout(align_items='center'))

##########################
# ITENS LIMIARIZAÇÃO     #
##########################

# Slider
slider_limiar = widgets.IntSlider(value=127, min=0, max=255, step=1, description='Limiar:',
                                  layout=widgets.Layout(justify_content='center'))
# Botão para segmentar
button_segmentar_limiar = widgets.Button(description='Segmentar')
# Atribuir função ao botão
button_segmentar_limiar.on_click(segmentacao_limiar)
# Botão fechar segmentação
button_fechar_segmentacao = widgets.Button(description='Fechar')
# Atribuir função ao botão
button_fechar_segmentacao.on_click(fechar_segmentacao)

box_slider_limiar = widgets.Box([slider_limiar],
                             layout=widgets.Layout(display='flex',justify_content='center',margin='0px'))
# Box para organizar os controles
box_buttons_seg_limiar = widgets.Box([button_segmentar_limiar,button_mascara,button_fechar_segmentacao],
                          layout=widgets.Layout(justify_content='center',margin='0px'))

# Box para os itens de Limiarização
box_limiar = widgets.VBox([box_slider_limiar, box_buttons_seg_limiar],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

#############################
# ITENS LIMIARIZAÇÃO OTSU   #
#############################
t = None
# Texto para indicar o limiar de Otsu
texto_otsu = widgets.Label(value=f'Limiar de Otsu = {t}')
# Botão para segmentar
button_segmentar_otsu = widgets.Button(description='Segmentar')
# Atribuir função ao botão
button_segmentar_otsu.on_click(segmentacao_otsu)
# Botão fechar segmentação
button_fechar_segmentacao = widgets.Button(description='Fechar')
# Atribuir função ao botão
button_fechar_segmentacao.on_click(fechar_segmentacao)

# Box para organizar os controles
box_buttons_seg_otsu = widgets.Box([button_segmentar_otsu,button_mascara,button_fechar_segmentacao],
                                   layout=widgets.Layout(justify_content='center',margin='0px'))
box_texto_otsu = widgets.Box([texto_otsu],
                             layout=widgets.Layout(display='flex',justify_content='center',margin='0px'))
box_texto_otsu.layout.display = 'none'  # Inicialmente oculto
# Box para os itens de Limiarização Otsu
box_otsu = widgets.VBox([box_buttons_seg_otsu,box_texto_otsu],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

#############################
# ITENS DETECCAO BORDAS     #
#############################
# Slider
slider_canny = widgets.FloatSlider(value=0.5, min=0, max=1.0, step=0.1,
                                   description='Proporção do Limiar Inferior:',
                                   style={'description_width': 'initial'},
                                   layout=widgets.Layout(width='400px'))  # Ajusta a largura total do slider)
# Botão para segmentar
button_segmentar_canny = widgets.Button(description='Segmentar')
# Atribuir função ao botão
button_segmentar_canny.on_click(segmentacao_canny)
# Botão fechar segmentação
button_fechar_segmentacao = widgets.Button(description='Fechar')
# Atribuir função ao botão
button_fechar_segmentacao.on_click(fechar_segmentacao)
# Texto para indicar o número de contornos
texto_canny = widgets.Label(value=f'Número de contornos:')


# Box para organizar os controles
box_buttons_seg_canny = widgets.Box([button_segmentar_canny,button_mascara,button_fechar_segmentacao],
                                   layout=widgets.Layout(justify_content='center',margin='0px'))
box_slider_canny = widgets.Box([slider_canny],
                             layout=widgets.Layout(justify_content='center',margin='0px'))
box_texto_canny = widgets.Box([texto_canny],
                             layout=widgets.Layout(display='flex',justify_content='center',margin='0px'))
box_texto_canny.layout.display = 'none'  # Inicialmente oculto

# Box para os itens de Limiarização
box_canny = widgets.VBox([box_slider_canny,box_buttons_seg_canny,box_texto_canny],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

################################
# ITENS DETECCAO WATERSHED     #
################################

# Botão para segmentar
button_segmentar_watershed = widgets.Button(description='Segmentar')
# Atribuir função ao botão
button_segmentar_watershed.on_click(segmentacao_watershed)
# Botão fechar segmentação
button_fechar_segmentacao = widgets.Button(description='Fechar')
# Atribuir função ao botão
button_fechar_segmentacao.on_click(fechar_segmentacao)


# Box para organizar os controles
box_buttons_seg_watershed = widgets.Box([button_segmentar_watershed,button_mascara,button_fechar_segmentacao],
                                   layout=widgets.Layout(justify_content='center',margin='0px'))

# Box para os itens de Limiarização
box_watershed = widgets.VBox([box_buttons_seg_watershed],
                               layout=widgets.Layout(justify_content='center',margin='0px'))

# Criar subabas para a aba segmentação
subtab_segmentacao = widgets.Tab(
    children=[widgets.VBox([box_hsv]),
              widgets.VBox([box_limiar]),
              widgets.VBox([box_otsu]),
              widgets.VBox([box_canny]),
              widgets.VBox([box_watershed])
    ],
    layout=widgets.Layout(min_width='700px', min_height='40px')  # Definir tamanho mínimo
)
subtab_segmentacao.set_title(0, 'Cor')
subtab_segmentacao.set_title(1, 'Limiarização')
subtab_segmentacao.set_title(2, 'Limiarização Otsu')
subtab_segmentacao.set_title(3, 'Detecção de Bordas')
subtab_segmentacao.set_title(4, 'Watershed')


# Definir o conteúdo de cada subaba
tab.children = [
    widgets.HBox([dropdown_arquivos,button_abrir, button_fechar],
                 layout=widgets.Layout(align_items='center', justify_content='center')),
    widgets.HBox([subtab_imagem],
                 layout=widgets.Layout(align_items='center', justify_content='center')),
    widgets.HBox([subtab_filtros],
                 layout=widgets.Layout(align_items='center', justify_content='center')),
    widgets.HBox([subtab_aprimoramentos],
                 layout=widgets.Layout(align_items='center', justify_content='center')),
    widgets.HBox([subtab_segmentacao],
                 layout=widgets.Layout(align_items='center', justify_content='center'))
]


################ FIM DO SCRIPT ########################
```
